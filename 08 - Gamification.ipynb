        sortItems.forEach(function(item) {
            item.addEventListener('dragstart', function(e) {
                draggedId = this.dataset.id;
                this.style.opacity = '0.4';
                e.dataTransfer.effectAllowed = 'move';
            });
            
            item.addEventListener('dragend', function() {
                this.style.opacity = '1';
                draggedId = null;
            });
        });
        
        bucketEls.forEach(function(bucket) {
            bucket.addEventListener('dragover', function(e) {
                e.preventDefault();
                this.style.background = '#f0f7ff';
                this.style.borderStyle = 'solid';
            });
            
            bucket.addEventListener('dragleave', function() {
                this.style.background = '#fff';
                this.style.borderStyle = 'dashed';
            });
            
            bucket.addEventListener('drop', function(e) {
                e.preventDefault();
                this.style.background = '#fff';
                this.style.borderStyle = 'dashed';
                
                if (draggedId) {
                    var targetBucket = this.dataset.bucket;
                    
                    // Remove from unsorted
                    unsorted = unsorted.filter(function(item) { return item.id !== draggedId; });
                    
                    // Remove from other buckets
                    for (var key in buckets) {
                        buckets[key].items = buckets[key].items.filter(function(id) { return id !== draggedId; });
                    }
                    
                    // Add to target bucket
                    buckets[targetBucket].items.push(draggedId);
                    showResults = false;
                    render();
                }
            });
        });
        
        checkBtn.addEventListener('click', function() {
            if (unsorted.length === 0) {
                showResults = true;
                render();
            }
        });
        
        resetBtn.addEventListener('click', function() {
            init();
        });
    }
    
    if (typeof NodeList.prototype.forEach !== 'function') {
        NodeList.prototype.forEach = Array.prototype.forEach;
    }
    
    init();
})();

        timeLeft = TIME_PER_QUESTION;
        gameState = 'ready';
        selectedAnswer = null;
        shuffledQuestions = shuffleArray(questions).slice(0, 10);
        currentOptions = [];
        if (timer) clearInterval(timer);
        timer = null;
        render();
    }
    
    function startGame() {
        gameState = 'playing';
        setupQuestion();
        startTimer();
        render();
    }
    
    function setupQuestion() {
        var q = shuffledQuestions[currentIndex];
        currentOptions = shuffleArray([q.a].concat(q.wrong));
        selectedAnswer = null;
        timeLeft = TIME_PER_QUESTION;
    }

        if (timer) clearInterval(timer);
        timer = setInterval(function() {
            timeLeft--;
            if (timeLeft <= 0) {
                timeUp();
            }

    }
    
    function stopTimer() {
        if (timer) {
            clearInterval(timer);
            timer = null;
        }
    }
    
    function timeUp() {
        stopTimer();
        selectedAnswer = '__timeout__';
        streak = 0;
        gameState = 'answered';
        render();
        setTimeout(nextQuestion, 1500);
    }
    
    function selectAnswer(answer) {
        if (gameState !== 'playing') return;
        stopTimer();
        selectedAnswer = answer;
        var q = shuffledQuestions[currentIndex];
        
        if (answer === q.a) {
            score++;

        }

    }
    
    function nextQuestion() {
        currentIndex++;
        if (currentIndex >= shuffledQuestions.length) {
            gameState = 'finished';

    }

## Asteroids Game Widget

A classic arcade-style Asteroids game implemented in HTML5 Canvas. Features ship rotation and thrust physics, asteroid splitting mechanics, particle effects, score tracking, lives system, and progressive difficulty levels. Ideal for gamification breaks, reward mechanics after completing modules, or demonstrating Canvas-based interactive content in notebooks. Controls: Arrow keys or WASD to move, SPACE to shoot.

<canvas id="aC" width="560" height="380" tabindex="0" style="display:block;margin:0 auto;border-radius:6px;outline:none;"></canvas>
<div style="text-align:center;margin-top:8px;font-family:sans-serif;font-size:0.85em;color:#666;">⬅️➡️ or A/D Rotate • ⬆️ or W Thrust • SPACE Shoot • Click canvas to focus</div>

<script>
(function() {
    var W = 560, H = 380;
    var canvas = document.getElementById('aC');
    var ctx = canvas.getContext('2d');
    var ship, asteroids, bullets, particles;
    var score = 0, lives = 3, level = 1;
    var gameState = 'ready';
    var keys = {};
    var respawnTimer = 0;
    
    function init() {
        ship = { x: W/2, y: H/2, angle: -Math.PI/2, vx: 0, vy: 0, thrust: false };
        asteroids = []; bullets = []; particles = [];
        score = 0; lives = 3; level = 1; gameState = 'ready';
        spawnAsteroids(4);
    }
    
    function spawnAsteroids(count) {
        for (var i = 0; i < count; i++) {
            var x, y;
            do { x = Math.random() * W; y = Math.random() * H; } while (dist(x, y, ship.x, ship.y) < 120);
            var angle = Math.random() * Math.PI * 2;
            var speed = 1 + Math.random() * 1.5;
            asteroids.push({ x: x, y: y, vx: Math.cos(angle) * speed, vy: Math.sin(angle) * speed, size: 3, radius: 35, rotation: Math.random() * Math.PI * 2, rotSpeed: (Math.random() - 0.5) * 0.05 });
        }
    }
    
    function dist(x1, y1, x2, y2) { return Math.sqrt((x2-x1)*(x2-x1) + (y2-y1)*(y2-y1)); }
    function wrap(obj) { if (obj.x < 0) obj.x = W; if (obj.x > W) obj.x = 0; if (obj.y < 0) obj.y = H; if (obj.y > H) obj.y = 0; }
    
    function createParticles(x, y, count, color) {
        for (var i = 0; i < count; i++) {
            var angle = Math.random() * Math.PI * 2;
            var speed = Math.random() * 3 + 1;
            particles.push({ x: x, y: y, vx: Math.cos(angle) * speed, vy: Math.sin(angle) * speed, life: 25, color: color || '#fff' });
        }
    }
    
    function update() {
        if (gameState !== 'playing' && gameState !== 'dead') return;
        if (gameState === 'playing') {
            if (keys['ArrowLeft'] || keys['a'] || keys['A']) ship.angle -= 0.07;
            if (keys['ArrowRight'] || keys['d'] || keys['D']) ship.angle += 0.07;
            ship.thrust = keys['ArrowUp'] || keys['w'] || keys['W'];
            if (ship.thrust) { ship.vx += Math.cos(ship.angle) * 0.15; ship.vy += Math.sin(ship.angle) * 0.15; }
            ship.vx *= 0.99; ship.vy *= 0.99;
            var speed = Math.sqrt(ship.vx*ship.vx + ship.vy*ship.vy);
            if (speed > 8) { ship.vx = (ship.vx/speed) * 8; ship.vy = (ship.vy/speed) * 8; }
            ship.x += ship.vx; ship.y += ship.vy; wrap(ship);
        }
        if (gameState === 'dead') { respawnTimer--; if (respawnTimer <= 0) { ship.x = W/2; ship.y = H/2; ship.vx = 0; ship.vy = 0; ship.angle = -Math.PI/2; gameState = 'playing'; } }
        for (var i = bullets.length - 1; i >= 0; i--) { bullets[i].x += bullets[i].vx; bullets[i].y += bullets[i].vy; bullets[i].life--; wrap(bullets[i]); if (bullets[i].life <= 0) bullets.splice(i, 1); }
        for (var j = 0; j < asteroids.length; j++) { asteroids[j].x += asteroids[j].vx; asteroids[j].y += asteroids[j].vy; asteroids[j].rotation += asteroids[j].rotSpeed; wrap(asteroids[j]); }
        for (var p = particles.length - 1; p >= 0; p--) { particles[p].x += particles[p].vx; particles[p].y += particles[p].vy; particles[p].life--; if (particles[p].life <= 0) particles.splice(p, 1); }
        for (var bi = bullets.length - 1; bi >= 0; bi--) {
            for (var ai = asteroids.length - 1; ai >= 0; ai--) {
                if (dist(bullets[bi].x, bullets[bi].y, asteroids[ai].x, asteroids[ai].y) < asteroids[ai].radius) {
                    var ast = asteroids[ai]; bullets.splice(bi, 1); score += (4 - ast.size) * 20; createParticles(ast.x, ast.y, 8, '#888');
                    if (ast.size > 1) { for (var k = 0; k < 2; k++) { var na = Math.random() * Math.PI * 2; var ns = 1.5 + Math.random() * 2; asteroids.push({ x: ast.x, y: ast.y, vx: Math.cos(na) * ns, vy: Math.sin(na) * ns, size: ast.size - 1, radius: ast.radius * 0.55, rotation: Math.random() * Math.PI * 2, rotSpeed: (Math.random() - 0.5) * 0.08 }); } }
                    asteroids.splice(ai, 1); break;
                }
            }
        }
        if (gameState === 'playing') { for (var si = 0; si < asteroids.length; si++) { if (dist(ship.x, ship.y, asteroids[si].x, asteroids[si].y) < asteroids[si].radius + 8) { createParticles(ship.x, ship.y, 15, '#4fc3f7'); lives--; if (lives <= 0) gameState = 'gameover'; else { gameState = 'dead'; respawnTimer = 90; } break; } } }
        if (asteroids.length === 0 && gameState === 'playing') { level++; spawnAsteroids(3 + level); }
    }
    
    function shoot() { if (gameState !== 'playing' || bullets.length >= 5) return; bullets.push({ x: ship.x + Math.cos(ship.angle) * 12, y: ship.y + Math.sin(ship.angle) * 12, vx: Math.cos(ship.angle) * 10, vy: Math.sin(ship.angle) * 10, life: 45 }); }
    
    function draw() {
        ctx.fillStyle = '#0a0a1a'; ctx.fillRect(0, 0, W, H);
        particles.forEach(function(p) { ctx.fillStyle = p.color; ctx.globalAlpha = p.life / 25; ctx.beginPath(); ctx.arc(p.x, p.y, 2, 0, Math.PI * 2); ctx.fill(); }); ctx.globalAlpha = 1;
        ctx.strokeStyle = '#777'; ctx.lineWidth = 2;
        asteroids.forEach(function(a) { ctx.save(); ctx.translate(a.x, a.y); ctx.rotate(a.rotation); ctx.beginPath(); var v = a.size === 3 ? 7 : 6; for (var i = 0; i < v; i++) { var ang = (i / v) * Math.PI * 2; var r = a.radius * (0.8 + Math.sin(i * 3) * 0.2); if (i === 0) ctx.moveTo(Math.cos(ang) * r, Math.sin(ang) * r); else ctx.lineTo(Math.cos(ang) * r, Math.sin(ang) * r); } ctx.closePath(); ctx.stroke(); ctx.restore(); });
        ctx.fillStyle = '#4fc3f7'; bullets.forEach(function(b) { ctx.beginPath(); ctx.arc(b.x, b.y, 3, 0, Math.PI * 2); ctx.fill(); });
        if (gameState === 'playing' || gameState === 'ready') { ctx.save(); ctx.translate(ship.x, ship.y); ctx.rotate(ship.angle); if (ship.thrust && gameState === 'playing') { ctx.fillStyle = '#ff6b35'; ctx.beginPath(); ctx.moveTo(-10, -5); ctx.lineTo(-18 - Math.random() * 6, 0); ctx.lineTo(-10, 5); ctx.fill(); } ctx.strokeStyle = '#4fc3f7'; ctx.lineWidth = 2; ctx.beginPath(); ctx.moveTo(14, 0); ctx.lineTo(-10, -8); ctx.lineTo(-6, 0); ctx.lineTo(-10, 8); ctx.closePath(); ctx.stroke(); ctx.restore(); }
        ctx.fillStyle = '#fff'; ctx.font = 'bold 16px monospace'; ctx.textAlign = 'left'; ctx.fillText('SCORE: ' + score + '   LIVES: ' + lives + '   LEVEL: ' + level, 10, 22);
        ctx.textAlign = 'center';
        if (gameState === 'ready') { ctx.fillStyle = 'rgba(0,0,0,0.6)'; ctx.fillRect(0, 0, W, H); ctx.fillStyle = '#4fc3f7'; ctx.font = 'bold 32px monospace'; ctx.fillText('ASTEROIDS', W/2, H/2 - 30); ctx.fillStyle = '#fff'; ctx.font = '14px sans-serif'; ctx.fillText('Arrows/WASD to move • SPACE to shoot', W/2, H/2 + 5); ctx.fillStyle = '#888'; ctx.fillText('Press SPACE to start', W/2, H/2 + 35); }
        if (gameState === 'gameover') { ctx.fillStyle = 'rgba(0,0,0,0.7)'; ctx.fillRect(0, 0, W, H); ctx.fillStyle = '#f44336'; ctx.font = 'bold 32px monospace'; ctx.fillText('GAME OVER', W/2, H/2 - 10); ctx.fillStyle = '#fff'; ctx.font = '18px sans-serif'; ctx.fillText('Score: ' + score, W/2, H/2 + 25); ctx.fillStyle = '#888'; ctx.font = '14px sans-serif'; ctx.fillText('Press SPACE to restart', W/2, H/2 + 55); }
        if (gameState === 'dead') { ctx.fillStyle = '#f44336'; ctx.font = 'bold 20px sans-serif'; ctx.fillText('GET READY...', W/2, H/2); }
    }
    
    function loop() { update(); draw(); requestAnimationFrame(loop); }
    function start() { if (gameState === 'ready') gameState = 'playing'; else if (gameState === 'gameover') { init(); gameState = 'playing'; } }
    
    canvas.addEventListener('keydown', function(e) { keys[e.key] = true; if (e.key === ' ') { e.preventDefault(); if (gameState === 'ready' || gameState === 'gameover') start(); else shoot(); } });
    canvas.addEventListener('keyup', function(e) { keys[e.key] = false; });
    canvas.addEventListener('click', function() { canvas.focus(); });
    
    canvas.focus();
    init();
    loop();
})();
