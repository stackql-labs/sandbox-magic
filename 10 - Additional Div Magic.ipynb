# Additional Div Magic

A collection of advanced interactive HTML/CSS/JS patterns for `%md-sandbox` cells. Each example is generalized with placeholder content — adapt the data, colors, and labels to fit your use case.

## 1. Interactive Accordion Cards

A row of selectable cards that expand an accordion panel with structured details. Useful for comparing permission levels, tiers, pricing plans, or any set of mutually exclusive options.

**How it works:**
- Each card is a clickable `<span>` with an `onclick` handler
- Clicking a card toggles an expandable detail panel below with animated open/close
- Card data is stored in a JavaScript array — add/remove items by editing the `ITEMS` array
- Each item supports `can` (positive attributes), `cannot` (negative attributes), and a `role` (summary)
- Clicking the active card again collapses the panel

var accCurrent = null;

function selectAccItem(id) {
  var wrap = document.getElementById('acc-detail-wrap');
  var card = document.getElementById('acc-detail-card');
  var p = ITEMS[id];

  document.querySelectorAll('.acc-box').forEach(function(b) {
    b.classList.toggle('active', parseInt(b.dataset.id) === id && accCurrent !== id);
  });

  if (accCurrent === id) {
    wrap.classList.remove('open');
    accCurrent = null;
    return;
  }

  accCurrent = id;

  var canHtml = p.can.map(function(c) { return '<li style="margin-bottom:4px;">' + c + '</li>'; }).join('');

  var cannotHtml = '';
  if (p.cannot.length > 0) {

      + p.cannot.map(function(c) { return '<li style="margin-bottom:4px;">' + c + '</li>'; }).join('')
      + '</ul></div>';
  }

  card.style.borderTopColor = p.color;
  card.innerHTML =
    '<div style="font-size:18.5pt;font-weight:700;margin-bottom:12px;color:#0b2026;">' + p.title + '</div>'

## 2. Flip Cards

Cards that reveal additional detail on hover using a 3D flip animation. Great for comparing two (or more) options side-by-side where each has a summary front and detailed back.

**How it works:**
- Uses CSS `perspective` and `transform: rotateY(180deg)` for the 3D flip effect
- `backface-visibility: hidden` ensures only the visible face is shown
- Front face shows a summary with an icon; back face shows a bullet list and a callout
- Cards are laid out in a flex container — add more cards by duplicating the `flip-card-*` div
- The flip is triggered purely by CSS `:hover` — no JavaScript required

## 3. Category Grid Cards

A 2x2 (or any) grid of color-coded category cards, each with a title and bullet list. Ideal for grouping related items by theme — features by domain, resources by type, tools by function, etc.

**How it works:**
- Uses CSS Grid (`display: grid; grid-template-columns: 1fr 1fr`) for a responsive 2-column layout
- Each card has a distinct background/border color scheme for visual separation
- Titles use HTML entity icons (no images needed) — swap the `&#...;` codes for different icons
- Easily extend to 3 columns (`1fr 1fr 1fr`) or add more rows by adding more `<div>` blocks

## 4. Gradient Info Cards

A row of cards with gradient backgrounds, each highlighting a key theme or pillar. Works well for executive summaries, value propositions, or the "three pillars" pattern.

**How it works:**
- Uses CSS Grid with 3 columns (`grid-template-columns: 1fr 1fr 1fr`)
- Each card has a `linear-gradient` background for visual depth
- Bold centered title with descriptive body text below
- Adjust to 2 or 4 columns by changing the `grid-template-columns` value
- The gradient direction (`135deg`) and color stops can be customized per card
